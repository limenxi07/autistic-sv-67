## Grid Adventure V1

### Overview

Grid Adventure V1 is, at its core, a grid-world environment enriched with varied mechanics. The input is a MxN grid (positive integers M, N). Each tile can be empty (representing floor) or contain one or more entities (agent, key, door, etc.), each with distinct behavior. The objective is collect any gems if present, before reaching the exit tile.

If you didn't skip the getting started section at the top of this notebook, you should already be familiar with the game.

### Constraints (Project Scope)

To keep the capstone project tractable, we enforce the following constraints:

- There is exactly 1 agent in the grid, with starting health **at most 5**.
- There is exactly 1 exit tile in the grid.
- The agent moves one tile per step, except under the speed (Boots) power-up, which doubles movement to two tiles per step.
- Each entity type has a fixed, known cost, damage, or reward (cost = -reward) where applicable:
  - Coins: reward 5.
  - Turn cost: 3 (Every turn incurs a cost of 3 (or reward -3) except for the final turn to win the game).
  - Lava: damage 2.
- Any key can be used to open any door, but each key can only be used once. 
- It is guaranteed that the initial grid will have an **equal number of keys and locked doors**
- It is also guaranteed that any doors would be **locked in the initial state.**
- There are **at most 5** gems in any grid.
- There are **at most 5** key-door pairs in any grid.
- There are **at most 5** pushable boxes in any grid.
- There can be **any number of** coins or lava.
- Power-up durations are fixed:
  - Boots (Speed): 5 steps.
  - Ghost (Phasing): 5 steps.
  - Shield: 5 uses.
- There are **at most 3** of each powerup type in the grid.
- There will be **no overlapping entities in the initial state** of the grid (i.e. each cell is empty, representing floor, or contains exactly 1 entity)
- The maximum grid size is in the graded test cases will be **15x15**

### Objective and Challenges

Your overarching objective is to build an agent that plays well across diverse levels. Key challenges:

- **Planning:** Navigate through mazes and interact with various entities within them. Formulate a sequence of actions that maximizes the total reward.
- **Image Interpretation:** The principal observation is a rendered top-down RGBA image. You must transform it into a representation suitable for search / planning.

We decompose the overarching problem into tasks to let you focus on one challenge at a time (see [Tasks](#Tasks)).

### Implementation

Below is a starter `Agent` with a minimal breadth-first search and image-parsing scaffold. Its simple classifier matches cells against representative public assets and is intended only as a baseline. You may modify it or replace it entirely.


````python
# Input states for Agent step function
from grid_adventure.grid import GridState
from grid_adventure.env import ImageObservation
from grid_adventure.constants import STEP_COST
from grid_adventure.entities import (
    AgentEntity,
    BoxEntity,
    CoinEntity,
    ExitEntity,
    GemEntity,
    KeyEntity,
    LavaEntity,
    LockedDoorEntity,
    PhasingPowerUpEntity,
    ShieldPowerUpEntity,
    SpeedPowerUpEntity,
    UnlockedDoorEntity,
    WallEntity,
)
from grid_adventure.movements import MOVEMENTS
from grid_adventure.objectives import OBJECTIVES
from grid_adventure.rendering import DEFAULT_ASSET_ROOT
from grid_universe.grid.entity import BaseEntity

# State steppers
from grid_adventure.step import Action
from grid_adventure.grid import step as grid_step

# Utility helpers
from collections import deque
from pathlib import Path
import numpy as np
from PIL import Image

class Agent:
    """Grid Adventure: Variant 1 agent template.

    This class is the single public interface that Coursemology will import and
    interact with when evaluating your submission. You should extend the
    internals (add helper classes / functions in other files if you wish) but
    MUST preserve:

    1. The class name: Agent
    2. The public method: step(self, state: GridState | ImageObservation) -> Action

    High-level lifecycle per environment tick:
        state  --->  step(...)  --->  Action

    The "state" object type depends on the task:
    - Task 1: A fully structured GridState instance.
    - Task 2: An ImageObservation dictionary whose primary observation is an RGBA image
      plus limited structured metadata in the 'info' sub-dict. In this case you
      typically perform perception to build (or approximate) an internal
      structured representation before planning.
    - Task 3: Input state could be either a GridState instance 
      or an ImageObservation dictionary

    Constraints:
    - Keep per-step latency small (single CPU, 1.5 GB RAM). Avoid O(W*H) scans of
      the full grid every step.
    - Determinism helps reproducibility; seed your own RNG if you add any
      random components.

    You may add __init__ parameters (with defaults) if needed for your own
    development, but the grader will instantiate Agent() with no arguments.
    """

    def __init__(self):
        """Initialize your agent.

        Put all one-time setup here (e.g., hardcoded ML model weights, 
        precomputing heuristic tables). Keep it fast and memory-light 
        to respect platform limits.
        """
        # Placeholder for any future initialization logic
        pass

    def step(self, state: GridState | ImageObservation) -> Action:
        """Return the next action given the current environment state.

        Parameters
        ----------
        state : GridState | ImageObservation
            - If a GridState instance (Tasks 1 and 3): you have direct, structured
              access to grid, entities, objective message, score, etc.
            - If an ImageObservation dictionary (Tasks 2 and 3): contains 'image' (HxWx4 RGBA uint8)
              plus 'info' sub-dictionary (agent stats, partial config, message).
              You likely need to parse the image into an internal representation.

        Returns
        -------
        Action
            A valid action from the Action enum. Must always return a member;
            never return None.
        """

        if not isinstance(state, GridState):
            state = self.parse(state)
        frontier = deque([(state, [])])
        visited = {str(state.grid)}

        while frontier:
            current_state, path = frontier.popleft()

            for action in Action:
                next_state = grid_step(current_state, action)
                if next_state.lose:
                    continue

                next_path = path + [action]
                if next_state.win:
                    return next_path[0]

                key = str(next_state.grid)
                if key in visited:
                    continue

                visited.add(key)
                frontier.append((next_state, next_path))

        return Action.WAIT
    
    def parse(self, observation: ImageObservation) -> GridState:
        """Convert an image observation into a GridState.

        Grid Play also calls this method to display the agent's reconstructed view.
        """
        info = observation["info"]
        config = info["config"]
        status = info["status"]
        width = config["width"]
        height = config["height"]

        gridstate = GridState(
            width=width,
            height=height,
            movement=MOVEMENTS["cardinal"],
            objective=OBJECTIVES["collect_gems_and_exit"],
            step_cost=STEP_COST,
            seed=None if config["seed"] == -1 else config["seed"],
            turn=status["turn"],
            score=status["score"],
            win=status["phase"] == "win",
            lose=status["phase"] == "lose",
            message=info["message"] or None,
            turn_limit=None if config["turn_limit"] == -1 else config["turn_limit"],
        )

        agent_count = 0
        for y, row in enumerate(self.split_cells(observation)):
            for x, cell_image in enumerate(row):
                entity = self.predict(cell_image)
                if entity is not None:
                    if isinstance(entity, AgentEntity):
                        agent_count += 1
                        entity.set_health(info["agent"]["health"]["current_health"])
                    gridstate.add((x, y), entity)

        if agent_count != 1:
            raise ValueError(
                f"predict() must identify exactly one AgentEntity, but found {agent_count}."
            )

        return gridstate

    def split_cells(self, observation: ImageObservation) -> list[list[np.ndarray]]:
        """Split an RGBA observation into rows of cell images."""
        image = observation["image"]
        config = observation["info"]["config"]
        width = config["width"]
        height = config["height"]
        image_height, image_width = image.shape[:2]
        cells = []

        for y in range(height):
            y_start = y * image_height // height
            y_end = (y + 1) * image_height // height
            row = []
            for x in range(width):
                x_start = x * image_width // width
                x_end = (x + 1) * image_width // width
                row.append(image[y_start:y_end, x_start:x_end, :])
            cells.append(row)

        return cells

    def predict(self, cell_image: np.ndarray) -> BaseEntity | None:
        """Return the entity most similar to one RGBA cell image, or None for floor."""
        entity_types = {
            "boots": SpeedPowerUpEntity,
            "box": BoxEntity,
            "coin": CoinEntity,
            "exit": ExitEntity,
            "gem": GemEntity,
            "ghost": PhasingPowerUpEntity,
            "human": AgentEntity,
            "key": KeyEntity,
            "lava": LavaEntity,
            "locked": LockedDoorEntity,
            "opened": UnlockedDoorEntity,
            "shield": ShieldPowerUpEntity,
            "wall": WallEntity,
        }
        image_size = (cell_image.shape[1], cell_image.shape[0])
        asset_root = Path(DEFAULT_ASSET_ROOT)

        with Image.open(asset_root / "floor" / "floor_1.png") as image:
            floor = image.convert("RGBA").resize(image_size)

        observed = cell_image.astype(np.int16)
        best_entity = None
        best_error = np.abs(observed - np.asarray(floor)).mean()

        for asset_name, entity_type in entity_types.items():
            asset_path = asset_root / asset_name / f"{asset_name}_1.png"
            with Image.open(asset_path) as image:
                asset = image.convert("RGBA").resize(image_size)

            candidate = np.asarray(Image.alpha_composite(floor, asset))
            error = np.abs(observed - candidate).mean()
            if error < best_error:
                best_error = error
                best_entity = entity_type

        return None if best_entity is None else best_entity()
    
    def info(self) -> dict[str, str]:
        """Return info about the agent.

        NOTE: This method is optional and intended for debugging in Grid Play only. 
        You do not need to implement it for Coursemology submission, but it can be 
        helpful for visualizing your agent's internal state during development. 
        Implementing this method will not affect grading.
        """
        # Optional: return info about the agent
        return {"name": "BFS Agent"}
````

### Evaluating Your Agent

We have provided utility functions in `utils.py` and the cell below for public smoke testing. They use the same per-run reward-based performance formula and similar result fields, but they do not reproduce hidden cases, three-case aggregation, or Coursemology's hard timeout. The local time limit is diagnostic: it marks a completed run as timed out but does not interrupt a running evaluation, so stop the cell manually if your agent takes too long. When comparing with grading, treat any local error or timeout as an effective performance of 0.

````python
from pprint import pprint

import pandas as pd

from grid_adventure.grid import to_state
from grid_adventure.rendering import DEFAULT_ASSET_ROOT
from public_levels.catalog import (
    PUBLIC_BOSS_LEVEL_BUILDERS,
    PUBLIC_BOSS_TIME_LIMIT,
    PUBLIC_GAMEPLAY_LEVEL_BUILDERS,
    PUBLIC_GAMEPLAY_TIME_LIMIT,
    PUBLIC_LEVEL_MAX_REWARDS,
    PUBLIC_TURN_LIMIT,
)
from public_levels.intro import build_level_boss, build_level_required_multiple
from utils import evaluate, create_env, evaluate_level, get_result_string


def _public_result_status(result: dict[str, object]) -> str:
    """Return one exclusive status, ordered from most to least severe."""
    if result["error"]:
        return "ERROR"
    if result["timeout"]:
        return "TIMEOUT"
    if result["win"]:
        return "WIN"
    if result["lose"]:
        return "LOSE"
    return "INCOMPLETE"


def evaluate_public_levels(
    agent_class: type[Agent],
    level_builders: dict[str, object],
    observation_type: str,
    *,
    task: str,
    seed: int | list[int],
    time_limit: int,
    verbose: bool = True,
) -> list[dict[str, object]]:
    """Evaluate one public level collection in one observation mode."""
    if observation_type not in {"gridstate", "image"}:
        raise ValueError("observation_type must be 'gridstate' or 'image'")
    if isinstance(seed, list) and len(seed) != len(level_builders):
        raise ValueError("A seed list must contain one seed per public level.")

    records = []
    for index, (level_label, builder) in enumerate(level_builders.items()):
        level_seed = seed if isinstance(seed, int) else seed[index]
        max_total_reward = PUBLIC_LEVEL_MAX_REWARDS[level_label]
        if verbose:
            print(
                f"Running [{index + 1}/{len(level_builders)}] "
                f"{level_label} ({observation_type})..."
            )
        result = evaluate_level(
            agent_class,
            builder,
            observation_type=observation_type,
            max_total_reward=max_total_reward,
            min_total_reward=max_total_reward * 1.5,
            turn_limit=PUBLIC_TURN_LIMIT,
            time_limit=time_limit,
            seed=level_seed,
        )
        status = _public_result_status(result)
        record = {
            "task": task,
            "public_level": level_label,
            "observation_type": observation_type,
            "status": status,
            **result,
            "effective_performance": (
                0.0 if status in {"ERROR", "TIMEOUT"} else result["performance"]
            ),
        }
        records.append(record)
        if verbose:
            print(f"Result: {get_result_string(record)}")

    return records


def run_task_1_public_tests(
    agent_class: type[Agent], *, seed: int | list[int] = 42, verbose: bool = True
) -> pd.DataFrame:
    return pd.DataFrame(
        evaluate_public_levels(
            agent_class,
            PUBLIC_GAMEPLAY_LEVEL_BUILDERS,
            "gridstate",
            task="Task 1",
            seed=seed,
            time_limit=PUBLIC_GAMEPLAY_TIME_LIMIT,
            verbose=verbose,
        )
    )


def run_task_2_public_tests(
    agent_class: type[Agent], *, seed: int | list[int] = 42, verbose: bool = True
) -> pd.DataFrame:
    return pd.DataFrame(
        evaluate_public_levels(
            agent_class,
            PUBLIC_GAMEPLAY_LEVEL_BUILDERS,
            "image",
            task="Task 2",
            seed=seed,
            time_limit=PUBLIC_GAMEPLAY_TIME_LIMIT,
            verbose=verbose,
        )
    )


def run_task_3_public_tests(
    agent_class: type[Agent],
    *,
    gridstate_seed: int | list[int] = 1,
    image_seed: int | list[int] = 2,
    verbose: bool = True,
) -> pd.DataFrame:
    records = evaluate_public_levels(
        agent_class,
        PUBLIC_BOSS_LEVEL_BUILDERS,
        "gridstate",
        task="Task 3",
        seed=gridstate_seed,
        time_limit=PUBLIC_BOSS_TIME_LIMIT,
        verbose=verbose,
    )
    records.extend(
        evaluate_public_levels(
            agent_class,
            PUBLIC_BOSS_LEVEL_BUILDERS,
            "image",
            task="Task 3",
            seed=image_seed,
            time_limit=PUBLIC_BOSS_TIME_LIMIT,
            verbose=verbose,
        )
    )
    return pd.DataFrame(records)


def run_all_public_tests(agent_class: type[Agent], *, verbose: bool = True) -> pd.DataFrame:
    """Run all 48 public smoke checks once; this is not a grade estimate."""
    return pd.concat(
        [
            run_task_1_public_tests(agent_class, verbose=verbose),
            run_task_2_public_tests(agent_class, verbose=verbose),
            run_task_3_public_tests(agent_class, verbose=verbose),
        ],
        ignore_index=True,
    )


def summarize_public_results(results: pd.DataFrame) -> pd.DataFrame:
    """Summarize public checks using zero performance for errors/timeouts."""
    def summary_row(scope: str, frame: pd.DataFrame) -> dict[str, object]:
        status_counts = frame["status"].value_counts()
        measured_runtime = frame.loc[
            frame["status"] != "ERROR", "runtime (sec)"
        ]
        return {
            "scope": scope,
            "checks": len(frame),
            "wins": int(status_counts.get("WIN", 0)),
            "losses": int(status_counts.get("LOSE", 0)),
            "timeouts": int(status_counts.get("TIMEOUT", 0)),
            "errors": int(status_counts.get("ERROR", 0)),
            "incomplete": int(status_counts.get("INCOMPLETE", 0)),
            "avg_effective_performance": round(
                float(frame["effective_performance"].mean()), 3
            ),
            "avg_runtime_sec": (
                round(float(measured_runtime.mean()), 3)
                if not measured_runtime.empty
                else None
            ),
        }

    rows = [
        summary_row(f"{task} ({observation_type})", frame)
        for (task, observation_type), frame in results.groupby(
            ["task", "observation_type"], sort=False
        )
    ]
    rows.append(summary_row("Overall", results))
    return pd.DataFrame(rows).set_index("scope")
````

### Task 1 - GridState Representation

### Hardcoding Your Trained ML Model
Because Coursemology disallows arbitrary auxiliary file uploads (e.g. separate `.pth` / `.pkl` weight files), any ML models your `Agent` depends on must ship **inline inside your submission**. To make this easy (and less error-prone than manually copying numbers), we provide two helper functions in `utils.py`:

- `generate_torch_loader_snippet(model, example_inputs=..., prefer='auto', compression='zlib')`
- `generate_sklearn_loader_snippet(model, compression='zlib')`

Each returns a fully self-contained Python source string defining a `get_model(...)` function. You copy-paste that function into your agent code. At runtime the function reconstructs the trained model from a Base64-encoded (optionally compressed) blob embedded directly in the source.

Below we show the end-to-end workflow for PyTorch and scikit-learn. We also keep (optional) manual hardcoding examples for learning.

Suppose you train a model on the following data:

````python
import numpy as np

X = np.array([[1, 1], [1, 2], [2, 2], [2, 3]])
y = np.array([1, 8, 2, 4])
````

#### PyTorch

Assume you have the following simple network (we'll train it briefly, then auto-generate a loader snippet):

````python
import torch
from torch import nn

torch.manual_seed(2109)

net = nn.Linear(2, 1)
````

Train the network:

````python
def train(net: nn.Module, X: np.ndarray, y: np.ndarray):
    X = torch.tensor(X, dtype=torch.float32)
    y = torch.tensor(y[:, None], dtype=torch.float32)
    criterion = nn.MSELoss()
    optimizer = torch.optim.Adam(net.parameters(), lr=0.01)
    for epoch in range(100):
        optimizer.zero_grad()
        output = net(X)
        loss = criterion(output, y)
        loss.backward()
        optimizer.step()

train(net, X, y)
````

**Option 1 (Recommended):** Automatically generate a self-contained loader snippet.

Run the following after training:

````python
# Auto-generate a loader snippet for the trained PyTorch model
from utils import generate_torch_loader_snippet

example_input = torch.tensor([[0.0, 0.0]], dtype=torch.float32)  # minimal example for tracing if needed
snippet = generate_torch_loader_snippet(net, example_inputs=example_input, prefer="auto", compression="zlib")
print(snippet)

# You would copy the FULL printed function definition (def get_model(...): ...) into your agent source code.
# Then inside Agent.__init__ you could do: self.model = get_model()
````

**Option 2 (Manual, Educational):** Manually copy weights into a freshly constructed model. This is more error-prone and not recommended for larger networks, but shown for completeness.


````python
# Set precision of printing numbers to 10 d.p.
torch.set_printoptions(precision=10)
print(net.state_dict())
````

Example output:
```python
OrderedDict({'weight': tensor([[0.8238971829, 0.8886168599]]), 'bias': tensor([0.5479648113])})
```

Notes:
- `OrderedDict` is from the `collections` module; `tensor` is a PyTorch tensor constructor.
- Ensure you exactly match shapes and dtypes.

To hardcode manually:

````python
from collections import OrderedDict
from torch import tensor

# declare the same model
model = nn.Linear(2, 1)

# Hardcode the weights from the previous step (input your values here)
coeffs = OrderedDict([('weight', tensor([[0.8238971829, 0.8886168599]])), ('bias', tensor([0.5479648113]))])

# Load the model with the hardcoded weights
model.load_state_dict(coeffs)

# The model now has the same weights as it was previously trained
X_tensor = torch.tensor(X, dtype=torch.float32)

# Check if the outputs are the same
torch.allclose(model(X_tensor), net(X_tensor))
````
