## Tutorial 2: Boxes, coins, and lava

Find an action sequence that completes the level with the highest possible score. A box can be pushed onto a coin, but it cannot be pushed onto lava.

````python
gridstate_2_2 = GridState(
    width=6,
    height=4,
    movement=MOVEMENTS["cardinal"],
    objective=OBJECTIVES["collect_gems_and_exit"],
    seed=1,
    step_cost=STEP_COST,
)

wall_positions = [
    (0, 0), (0, 3), (1, 0), (1, 3), (2, 3),
    (3, 2), (3, 3), (4, 2), (4, 3), (5, 2), (5, 3),
]
for position in wall_positions:
    gridstate_2_2.add(position, WallEntity())

gridstate_2_2.add((0, 1), AgentEntity())
gridstate_2_2.add((2, 1), BoxEntity())
gridstate_2_2.add((3, 1), CoinEntity())
gridstate_2_2.add((4, 1), LavaEntity())
gridstate_2_2.add((5, 1), ExitEntity())

display(renderer.render(to_state(gridstate_2_2)))
````

````python
# TODO: Complete the action list using Action.UP, Action.DOWN,
# Action.LEFT, Action.RIGHT, and Action.PICK_UP.
actions = []

result_2_2 = gridstate_2_2
for action in actions:
    result_2_2 = step(result_2_2, action)

display(renderer.render(to_state(result_2_2)))

if result_2_2.win:
    print("Congratulations! You solved the grid. Total score:", result_2_2.score)
    if result_2_2.score == -16:
        print("You achieved the highest possible score!")
    else:
        print("Try another route to reach the optimal score of -16.")
elif result_2_2.lose:
    print("The agent lost. Try another action sequence.")
else:
    print("The grid is not yet solved. Keep trying!")
````

## Tutorial 3: Gems, keys, doors, and shields

Complete the level. The score does not matter. The agent starts with 1 health point, so entering lava without an active shield causes a loss.

````python
gridstate_2_3 = GridState(
    width=9,
    height=3,
    movement=MOVEMENTS["cardinal"],
    objective=OBJECTIVES["collect_gems_and_exit"],
    seed=0,
    step_cost=STEP_COST,
)

agent = AgentEntity()
agent.set_health(1)
gridstate_2_3.add((0, 2), agent)

wall_positions = [
    (1, 0), (1, 2),
    (2, 0), (2, 2),
    (3, 0), (3, 2),
    (5, 0), (5, 2),
    (6, 0), (6, 2),
    (7, 0), (7, 2),
]
for position in wall_positions:
    gridstate_2_3.add(position, WallEntity())

gridstate_2_3.add((1, 1), ShieldPowerUpEntity())
gridstate_2_3.add((5, 1), LavaEntity())

gem_positions = [(4, 1), (8, 0)]
for position in gem_positions:
    gridstate_2_3.add(position, GemEntity())

gridstate_2_3.add((4, 0), KeyEntity())
gridstate_2_3.add((8, 1), LockedDoorEntity())
gridstate_2_3.add((8, 2), ExitEntity())

display(renderer.render(to_state(gridstate_2_3)))
````

````python
# TODO: Complete the action list using Action.UP, Action.DOWN,
# Action.LEFT, Action.RIGHT, Action.PICK_UP, and Action.USE_KEY.
actions = []

result_2_3 = gridstate_2_3
for action in actions:
    result_2_3 = step(result_2_3, action)

display(renderer.render(to_state(result_2_3)))

if result_2_3.win:
    print("Congratulations! You solved the grid. Total score:", result_2_3.score)
elif result_2_3.lose:
    print("The agent lost. Try another action sequence.")
else:
    print("The grid is not yet solved. Keep trying!")
````
